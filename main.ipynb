```python
from core.generators.player_generator import PlayerGenerator
gen = PlayerGenerator()
p = gen.generate(1)
print(p.full_name(), p.rating, p)
```

```python
import matplotlib.pyplot as plt
s = []
n = 1000
for i in range(n):
    p = gen.generate(i)
    age, rat, pot = p.age, p.rating, p.potential
    s.append((age, rat, pot))

for _ in range(3):
    plt.hist([s[i][_] for i in range(n)], bins=100, density=True)
    plt.title(['age', 'rating', 'potential'][_])
    plt.show()
```

```python
import argparse
from interface.cli.commands import cmd_generate_players, cmd_simulate

def main():
    parser = argparse.ArgumentParser(description='Football Simularoe CLI')
    subparsers = parser.add_subparsers(dest='command', required=True)

    gen_parser = subparsers.add_parser('generate', help='Сгенерировать игроков')
    gen_parser.add_argument('--count', type=int, default=100, help='Количество игроков')
    gen_parser.set_defaults(func=cmd_generate_players)

    sim_parser = subparsers.add_parser('sim', help='Симулировать матч')
    sim_parser.set_defaults(func=cmd_simulate)

    args = parser.parse_args()
    args.func(args)

if __name__ == '__main__':
    main()
```
