PLAYER

```python
from dataclasses import dataclass
from enum import Enum
from typing import Optional

class Position(Enum):
    GK = 'Goalkeeper'
    CB = 'Center Back'
    FB = 'Full Back'
    DM = 'Defense Midfielder'
    CM = 'Central Midfielder'
    AM = 'Attacking Midfielder'
    WG = 'Winger'
    ST = 'Striker'

@dataclass
class Player:
    id: int
    first_name: str
    last_name: str
    nation: str
    age: int
    position: Position
    rating: int
    potential: int
    team_id: Optional[int] = None

    def full_name(self) -> str:
        return f'{self.first_name} {self.last_name}'

```

TEAM

```python
from dataclasses import dataclass
from typing import List, Optional
from core.entities.player import Player

@dataclass
class Team:
    id: int
    name: str
    city: str
    stadium: str
    players: List[Player] = None

    def __post_init__(self):
        if  self.players is None:
            self.players = []
        
    def add_players(self, player: Player):
        self.players.append(player)
    
    def get_best_lineup(self) -> List[Player]:
        if  len(self.players) < 11:
            return self.players
        
        gk = [p for p in self.players if p.position.name == 'GK']
        df = [p for p in self.players if p.position.name in ('CB', 'FB')]
        mf = [p for p in self.players if p.position.name in ('DM', 'CM', 'AM')]
        str = [p for p in self.players if p.position.name in ('ST', 'WG')]

        gk.sort(key=lambda x: x.rating, reverse=True)
        df.sort(key=lambda x: x.rating, reverse=True)
        mf.sort(key=lambda x: x.rating, reverse=True)
        str.sort(key=lambda x: x.rating, reverse=True)

        lineup = []
        lineup.append(gk[0] if gk else None)
        lineup.extend(df[:4])
        lineup.extend(mf[:4])
        lineup.extend(str[:2])

        lineup = [p for p in lineup if p is not None]

        if  len(lineup) < 11:
            return self.players
        return lineup[:11]


    

```
