```python
import numpy as np
from typing import List, Tuple
from core.entities.player import Player, Position
from core.entities.team import Team

def team_strength(team: List[Player]) -> float:
    weights = {
        Position.GK: 1.3,
        Position.CB: 1.2,
        Position.FB: 1.0,
        Position.DM: 1.2,
        Position.CM: 1.1,
        Position.AM: 1.2, 
        Position.WG: 1.2,
        Position.ST: 1.3
    }

    total = 0.
    for player in team:
        total += player.rating * weights.get(player.position, 1.)
    return total / len(team)

def simulate_match(home_team: Team, away_team: Team) -> Tuple[int, int, List[Player], List[Player]]:
    home_lineup = home_team.get_best_lineup()
    away_lineup = away_team.get_best_lineup()
    
    home_str = team_strength(home_lineup)
    away_str = team_strength(away_lineup)

    base_lamdba = 2.5
    k = (home_str / away_str) ** 0.5
    home_lamdba = base_lamdba * k
    away_lamdba = base_lamdba / k

    home_lamdba = np.clip(home_lamdba, 0.3, 5.)
    away_lamdba = np.clip(away_lamdba, 0.3, 5.)

    home_goals = np.random.poisson(home_lamdba)
    away_goals = np.random.poisson(away_lamdba)

    return home_goals, away_goals, home_lineup, away_lineup
```
