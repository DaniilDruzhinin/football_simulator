PLAYER GENERATOR

```python
import random
import numpy as np
from core.entities.player import Player, Position
from core.utils.data_loader import load_last_names, load_first_names, load_nations

class PlayerGenerator:
    def __init__(self):
        self.first_names = load_first_names()
        self.last_names = load_last_names()
        self.nations = load_nations()
    
    def generate(self, player_id: int, team_id: int = None) -> Player:
        nation = random.choices(
            population= list(self.nations.keys()),
            weights= list(self.nations.values()))[0]
        
        first = random.choice(self.first_names.get(nation, ['Unknown']))
        last = random.choice(self.last_names.get(nation, ['Unknown']))
        age = int(np.clip(np.random.normal(24, 5), 16, 40)) 

        position = random.choice(list(Position))

        base_rating = self._base_rating_for_age(age)
        rating = int(np.clip(np.random.normal(base_rating, 8), 0, 100))

        potential_gap = max(5, int(np.random.gamma(shape=3, scale=5)))
        if  age > 30:
            potential_gap = max(0, potential_gap - (age - 30)*2)

        potential = min(99, rating + potential_gap)

        return Player(
            id=player_id,
            first_name=first,
            last_name=last,
            nation=nation,
            age=age,
            position=position,
            rating=rating,
            potential=potential,
            team_id=team_id
        )

    def _base_rating_for_age(self, age: int) -> float:
        if age < 21:
            return 40 + (age - 16) * 2
        elif age < 28:
            return 55 + (age - 21) * 3
        else:
            return 76 - (age - 28) * 2.5
```

TEAM GENERATOR

```python
import json
from pathlib import Path
from core.entities.team import Team
from core.utils.data_loader import load_teams

def load_teams_from_json():
    teams_data = load_teams()
    teams = []
    for td in teams_data:
        teams.append(Team(
            id= td['id'],
            name= td['name'],
            city= td['city'],
            stadium= td['stadium'],
        ))
    return teams
```
